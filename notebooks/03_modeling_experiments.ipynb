* `random_state=42` : Fixé globalement sur tous les modèles et les fractionnements pour garantir la reproductibilité mathématique exacte des résultats à chaque exécution.

* `max_features=2500` (TF-IDF) : Contrainte sémantique injectée pour limiter la dimension de la matrice textuelle des e-mails, évitant ainsi la saturation de la mémoire vive (RAM) de la machine.
